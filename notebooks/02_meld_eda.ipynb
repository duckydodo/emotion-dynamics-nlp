```python
for name, df in {
    "train": train_df,
    "dev": dev_df,
    "test": test_df
}.items():
    print(f"\n{name.upper()}")
    print(df.isnull().sum())
```

```python
for name, df in {
    "train": train_df,
    "dev": dev_df,
    "test": test_df
}.items():

    duplicates = df.duplicated(
        subset=["Dialogue_ID", "Utterance_ID"]
    ).sum()

    print(f"{name}: {duplicates} duplicate dialogue-turn IDs")
```

```python
for name, df in {
    "train": train_df,
    "dev": dev_df,
    "test": test_df
}.items():

    broken_dialogues = 0

    for dialogue_id, dialogue in df.groupby("Dialogue_ID"):
        ids = sorted(dialogue["Utterance_ID"].tolist())

        if ids != list(range(len(ids))):
            broken_dialogues += 1

    print(f"{name}: {broken_dialogues} dialogues with non-contiguous turns")
```

```python
splits = {
    "train": train_df,
    "dev": dev_df,
    "test": test_df
}

for name, df in splits.items():

    n_utterances = len(df)
    n_dialogues = df["Dialogue_ID"].nunique()
    n_speakers = df["Speaker"].nunique()

    print(
        f"{name:5} | "
        f"{n_dialogues:4,} dialogues | "
        f"{n_utterances:5,} utterances | "
        f"{n_speakers:3,} speakers"
    )
```

```python
total_dialogues = sum(
    df["Dialogue_ID"].nunique()
    for df in splits.values()
)

total_utterances = sum(len(df) for df in splits.values())

print(f"Total dialogues:  {total_dialogues:,}")
print(f"Total utterances: {total_utterances:,}")
```

```python
for name, df in splits.items():
    nonzero_start = []
    internal_gaps = []

    for dialogue_id, dialogue in df.groupby("Dialogue_ID"):
        ids = sorted(dialogue["Utterance_ID"].tolist())

        if ids[0] != 0:
            nonzero_start.append((dialogue_id, ids))

        expected = list(range(ids[0], ids[-1] + 1))

        if ids != expected:
            internal_gaps.append((dialogue_id, ids))

    print(f"\n{name.upper()}")
    print(f"Non-zero starts: {len(nonzero_start)}")
    print(f"Internal gaps:   {len(internal_gaps)}")
```

```python
for name, df in splits.items():
    print(f"\n{name.upper()}")

    shown = 0

    for dialogue_id, dialogue in df.groupby("Dialogue_ID"):
        ids = sorted(dialogue["Utterance_ID"].tolist())
        expected = list(range(ids[0], ids[-1] + 1))

        if ids != expected:
            print(f"Dialogue {dialogue_id}: {ids}")
            shown += 1

            if shown == 5:
                break
```

```python
train_ids = sorted(train_df["Dialogue_ID"].unique())

print("Min:", min(train_ids))
print("Max:", max(train_ids))
print("Unique:", len(train_ids))

missing_ids = sorted(
    set(range(min(train_ids), max(train_ids) + 1))
    - set(train_ids)
)

print("Missing IDs:", missing_ids)
```

```python
for name, df in splits.items():
    ids = sorted(df["Dialogue_ID"].unique())

    missing = sorted(
        set(range(min(ids), max(ids) + 1))
        - set(ids)
    )

    print(
        f"{name}: range {min(ids)}–{max(ids)}, "
        f"{len(ids)} unique, missing IDs = {missing}"
    )
```

```python
gap_sizes = []
missing_turns = 0
affected_dialogues = 0

for name, df in splits.items():
    split_affected = 0
    split_missing = 0

    for dialogue_id, dialogue in df.groupby("Dialogue_ID"):
        ids = sorted(dialogue["Utterance_ID"].tolist())

        dialogue_has_gap = False

        for current_id, next_id in zip(ids, ids[1:]):
            gap = next_id - current_id - 1

            if gap > 0:
                gap_sizes.append(gap)
                split_missing += gap
                dialogue_has_gap = True

        if dialogue_has_gap:
            split_affected += 1

    missing_turns += split_missing
    affected_dialogues += split_affected

    print(
        f"{name}: {split_affected} affected dialogues, "
        f"{split_missing} missing turn positions"
    )

print(f"\nTotal affected dialogues: {affected_dialogues}")
print(f"Total missing turn positions: {missing_turns}")
```

```python
from collections import Counter

print(Counter(gap_sizes))
```

```python
emotion_order = [
    "neutral",
    "joy",
    "surprise",
    "anger",
    "sadness",
    "disgust",
    "fear"
]

for name, df in splits.items():
    counts = df["Emotion"].value_counts()

    print(f"\n{name.upper()}")

    for emotion in emotion_order:
        count = counts.get(emotion, 0)
        percentage = count / len(df) * 100

        print(
            f"{emotion:10} "
            f"{count:>5,} "
            f"({percentage:5.2f}%)"
        )
```

```python
all_df = pd.concat(
    [
        df.assign(Split=name)
        for name, df in splits.items()
    ],
    ignore_index=True
)

counts = all_df["Emotion"].value_counts()

print("OVERALL")

for emotion in emotion_order:
    count = counts.get(emotion, 0)

    print(
        f"{emotion:10} "
        f"{count:>5,} "
        f"({count / len(all_df) * 100:5.2f}%)"
    )
```

```python
valid_pairs = []

for split_name, df in splits.items():
    for dialogue_id, dialogue in df.groupby("Dialogue_ID"):

        dialogue = dialogue.sort_values("Utterance_ID")

        rows = list(dialogue.itertuples())

        for current, nxt in zip(rows, rows[1:]):

            # Only genuinely adjacent turns
            if nxt.Utterance_ID != current.Utterance_ID + 1:
                continue

            valid_pairs.append({
                "split": split_name,
                "dialogue_id": dialogue_id,
                "current_emotion": current.Emotion,
                "next_emotion": nxt.Emotion,
                "current_speaker": current.Speaker,
                "next_speaker": nxt.Speaker,
            })

len(valid_pairs)
```

```python
sum(
    len(df) - df["Dialogue_ID"].nunique()
    for df in splits.values()
)
```

```python
import numpy as np
from collections import Counter

transition_counts = []
unique_emotion_counts = []
all_neutral = 0

for split_name, df in splits.items():
    for dialogue_id, dialogue in df.groupby("Dialogue_ID"):

        dialogue = dialogue.sort_values("Utterance_ID")

        emotions = dialogue["Emotion"].tolist()
        ids = dialogue["Utterance_ID"].tolist()

        if all(e == "neutral" for e in emotions):
            all_neutral += 1

        unique_emotion_counts.append(len(set(emotions)))

        changes = 0

        for i in range(len(emotions) - 1):
            if ids[i + 1] != ids[i] + 1:
                continue

            if emotions[i] != emotions[i + 1]:
                changes += 1

        transition_counts.append(changes)

transition_counts = np.array(transition_counts)

print(f"Total dialogues: {total_dialogues:,}")

print(
    f"All-neutral dialogues: {all_neutral:,} "
    f"({all_neutral / total_dialogues * 100:.2f}%)"
)

print(f"Mean changes/dialogue:   {transition_counts.mean():.2f}")
print(f"Median changes/dialogue: {np.median(transition_counts):.0f}")
print(f"Maximum changes:         {transition_counts.max()}")

print(
    f"Zero changes: "
    f"{(transition_counts == 0).sum():,} "
    f"({(transition_counts == 0).mean() * 100:.2f}%)"
)

print(
    f"At least one change: "
    f"{(transition_counts > 0).sum():,} "
    f"({(transition_counts > 0).mean() * 100:.2f}%)"
)
```

```python
diversity = Counter(unique_emotion_counts)

for n_unique, count in sorted(diversity.items()):
    print(
        f"{n_unique} unique emotion(s): "
        f"{count:,} dialogues "
        f"({count / total_dialogues * 100:.2f}%)"
    )
```

```python
distribution = Counter(transition_counts)

for n_changes, count in sorted(distribution.items()):
    print(
        f"{n_changes:2} changes: "
        f"{count:4,} dialogues "
        f"({count / total_dialogues * 100:5.2f}%)"
    )
```

```python
pairs_df = pd.DataFrame(valid_pairs)

total_forecasts = len(pairs_df)

same = (
    pairs_df["current_emotion"]
    == pairs_df["next_emotion"]
).sum()

changes = total_forecasts - same

print(f"Valid forecasting examples: {total_forecasts:,}")

print(
    f"Same-state: {same:,} "
    f"({same / total_forecasts * 100:.2f}%)"
)

print(
    f"Change:     {changes:,} "
    f"({changes / total_forecasts * 100:.2f}%)"
)
```

```python
target_counts = pairs_df["next_emotion"].value_counts()

print(target_counts)

majority_class = target_counts.idxmax()
majority_accuracy = target_counts.max() / total_forecasts

print(
    f"\nAlways {majority_class}: "
    f"{majority_accuracy * 100:.2f}%"
)

print(
    f"Predict current emotion again: "
    f"{same / total_forecasts * 100:.2f}%"
)
```

```python
same_speaker = (
    pairs_df["current_speaker"]
    == pairs_df["next_speaker"]
)

emotion_change = (
    pairs_df["current_emotion"]
    != pairs_df["next_emotion"]
)

print(
    f"Same-speaker adjacent pairs: "
    f"{same_speaker.sum():,} "
    f"({same_speaker.mean() * 100:.2f}%)"
)

print(
    f"Speaker-change adjacent pairs: "
    f"{(~same_speaker).sum():,} "
    f"{(~same_speaker).mean() * 100:.2f}%)"
)
```

```python
same_speaker_change_rate = emotion_change[same_speaker].mean()

other_speaker_change_rate = emotion_change[~same_speaker].mean()

print(
    f"Emotion change after same speaker: "
    f"{same_speaker_change_rate * 100:.2f}%"
)

print(
    f"Emotion change after speaker switch: "
    f"{other_speaker_change_rate * 100:.2f}%"
)
```

```python
speakers_per_dialogue = []

for split_name, df in splits.items():
    for dialogue_id, dialogue in df.groupby("Dialogue_ID"):
        speakers_per_dialogue.append(
            dialogue["Speaker"].nunique()
        )

speakers_per_dialogue = np.array(speakers_per_dialogue)

print(f"Mean speakers/dialogue:   {speakers_per_dialogue.mean():.2f}")
print(f"Median speakers/dialogue: {np.median(speakers_per_dialogue):.0f}")
print(f"Min:                      {speakers_per_dialogue.min()}")
print(f"Max:                      {speakers_per_dialogue.max()}")

print("\nDistribution:")

for n, count in sorted(Counter(speakers_per_dialogue).items()):
    print(
        f"{n} speaker(s): "
        f"{count:,} dialogues "
        f"({count / total_dialogues * 100:.2f}%)"
    )
```
